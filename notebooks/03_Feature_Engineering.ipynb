```python
import joblib
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import StandardScaler
```

```python
df = pd.read_csv("../data/air_quality_clean.csv")

df.head()
```

```python
print(df.isnull().sum())
```

```python
numerical_columns = [
    "pollutant_min",
    "pollutant_max",
    "pollutant_avg"
]

for column in numerical_columns:
    df[column] = df[column].fillna(df[column].median())
```

```python
print(df.isnull().sum())
```

```python
label_encoders = {}

categorical_columns = [
    "state",
    "city",
    "pollutant_id",
    "AQI_Bucket"
]

for column in categorical_columns:
    encoder = LabelEncoder()

    df[column] = encoder.fit_transform(df[column])

    label_encoders[column] = encoder
```

```python
joblib.dump(
    label_encoders,
    "../models/label_encoders.pkl"
)
```

```python
X_reg = df.drop(columns=["AQI", "AQI_Bucket"])

y_reg = df["AQI"]
```

```python
X_cls = df.drop(columns=["AQI", "AQI_Bucket"])

y_cls = df["AQI_Bucket"]
```

```python
X_train_reg, X_test_reg, y_train_reg, y_test_reg = train_test_split(
    X_reg,
    y_reg,
    test_size=0.2,
    random_state=42
)
```

```python
X_train_cls, X_test_cls, y_train_cls, y_test_cls = train_test_split(
    X_cls,
    y_cls,
    test_size=0.2,
    random_state=42
)
```

```python
scaler = StandardScaler()

X_train_reg_scaled = scaler.fit_transform(X_train_reg)

X_test_reg_scaled = scaler.transform(X_test_reg)
```

```python
X_train_cls_scaled = scaler.fit_transform(X_train_cls)

X_test_cls_scaled = scaler.transform(X_test_cls)
```

```python
joblib.dump(
    scaler,
    "../models/scaler.pkl"
)
```

```python
joblib.dump(
    (X_train_reg, X_test_reg, y_train_reg, y_test_reg),
    "../models/regression_data.pkl"
)

joblib.dump(
    (X_train_cls, X_test_cls, y_train_cls, y_test_cls),
    "../models/classification_data.pkl"
)
```
